```python
import numpy as np
import soundfile as sf
import math
import os

def aud_feats_no_vad_3Channel(infile, outfile, fs):
    """
    Multi-Channel FDLP feature extraction without VAD.

    Parameters
    ----------
    infile : str
        Path to input audio file (.wav)
    outfile : str
        Path to output feature file (.htk or binary)
    fs : int or str
        Sampling frequency in Hz

    Notes
    -----
    - Reads a multi-channel audio file.
    - Calls generate_env_feats() to compute FDLP envelope features.
    - Applies logarithmic compression.
    - Writes the output in HTK-compatible format.
    """

    # Convert fs to integer if string
    if isinstance(fs, str):
        fs = int(fs)

    # Check input file
    if not os.path.exists(infile):
        raise FileNotFoundError(f"Input file '{infile}' not found.")

    # Read the audio
    y1, file_fs = sf.read(infile)
    if fs != file_fs:
        print(f"⚠️ Warning: Sampling rate mismatch ({file_fs} in file, expected {fs})")

    # Ensure it's a 2D array: (samples, channels)
    if y1.ndim == 1:
        y1 = y1[:, np.newaxis]

    # Transpose to match MATLAB (channels × samples)
    samples = (y1.T * (2**15)).astype(np.float64)
    nochan = samples.shape[0]

    # ---- FDLP Envelope Feature Extraction ----
    # Placeholder for your FDLP feature generation function
    ##bp()
    cepstra = generate_env_feats(samples, fs, nochan)

    # Log compression
    cepstra = np.log(np.maximum(cepstra, 1e-12))  # avoid log(0)

    #bp()

    # ---- Write to HTK format ----
    #writehtkf_new(outfile, cepstra, 100000.0, 8267)

    print(f"✅ FDLP features written to {outfile}")
    return cepstra

```

```python
import numpy as np
from scipy.signal import detrend

def generate_env_feats(A, sr, nochan):
    """
    Generate FDLP envelope-based cepstral features.

    Parameters
    ----------
    A : np.ndarray
        Multi-channel input signal array of shape (nochan, samples)
    sr : int
        Sampling rate (Hz)
    nochan : int
        Number of channels

    Returns
    -------
    cepstra : np.ndarray
        FDLP cepstral features [num_ceps × num_frames]
    """

    # Parameters (from MATLAB)
    do_gain_norm = True
    ceps = 14
    do_delta = True
    do_c0 = True
    ##bp()

    flen = int(0.025 * sr)   # 25 ms
    fhop = int(0.010 * sr)   # 10 ms
    total_samples = A.shape[1]

    fnum = int(np.floor((total_samples - flen) / fhop) + 1)
    send = int((fnum - 1) * fhop + flen)
    A = A[:, :send]

    fdlpwin = min(2 * sr, send)  # 2 s FDLP window
    fdlpolap = int(0.020 * sr)   # 20 ms overlap

    # Frame the input
    X = []
    add_samp = None
    for j in range(nochan):
        x_frames, add_samp = frame_new(A[j, :], fdlpwin, fdlpolap)
        X.append(x_frames)
    X = np.stack(X, axis=0)  # shape: (nochan, frame_len, num_frames)

    cepstra = []

    for i in range(X.shape[2]):
        flagbrk = False
        x = X[:, :, i]

        # Handle the last chunk append
        if X.shape[2] == 1:
            x = X[:, :-add_samp, i] if add_samp else X[:, :, i]

        if i == X.shape[2] - 2:
            concat = np.concatenate(
                (X[:, :, i], X[:, fdlpolap+1:-add_samp, i+1]), axis=1)
            x = concat
            flagbrk = True

        # Dither + Detrend
        ##bp()
        #x = ditherit(x, mode='bit') Commented by Anurenjan on 30-10-2025,should be double checked
        #should be uncommented later
        x = detrend(x, axis=-1, type='constant')

        # Skip very short segments
        if x.shape[1] < 400:
            print(f"⚠️ Warning: frame {i} too short — skipping.")
            continue

        flag_delta = True

        ##bp()

        temp = fdlp_env_comp_100hz(
            x, sr, ceps, flag_delta, do_gain_norm, nochan)
        #bp()

        cepstra.append(temp)

        if flagbrk:
            break

    if len(cepstra) == 0:
        return np.zeros((ceps, 1))

    cepstra = np.concatenate(cepstra, axis=1)
    return cepstra



```

```python
import numpy as np

def frame_new(x, FRAME, R):
    """
    Frame the input signal into overlapping segments.

    Parameters
    ----------
    x : np.ndarray
        Input 1D signal
    FRAME : int
        Frame length (number of samples per frame)
    R : int
        Overlap in samples between consecutive frames

    Returns
    -------
    xtr : np.ndarray
        2D array of shape (FRAME, NB_FRAMES), containing overlapping frames
    add_samp : int
        Number of padding samples added to complete the last frame
    """

    # Ensure x is a 1D numpy array
    x = np.asarray(x).flatten()

    # Compute number of frames
    NB_FR = int(np.ceil((len(x) - FRAME) / (FRAME - R) + 1))
    req_len = int(FRAME + (NB_FR - 1) * (FRAME - R))

    add_samp = req_len - len(x)

    # Dither + zero-padding for the missing samples
    if add_samp > 0:
        pad = np.zeros(add_samp)
        pad = ditherit_2(pad, amp=1.0, mode='bit')
        x = np.concatenate((x, pad))

    if NB_FR > 0:
        xtr = enframe(x, FRAME, FRAME - R).T  # shape (FRAME, NB_FR)
    else:
        raise ValueError("Cannot create frames — check frame and overlap values.")

    return xtr, add_samp



```

```python
import numpy as np

def enframe(x, win, inc=None):
    """
    Split a 1D signal into overlapping frames (rows of output matrix).

    Parameters
    ----------
    x : np.ndarray
        Input signal (1D array).
    win : int or np.ndarray
        If int, frame length. If 1D array, used as window and frame length = len(win)
    inc : int, optional
        Increment (hop) between frames. Defaults to frame length.

    Returns
    -------
    f : np.ndarray
        2D array of shape (num_frames, frame_length) where each row is a frame.
    """

    x = np.asarray(x).flatten()  # Ensure 1D
    nx = len(x)

    # Determine frame length
    if np.isscalar(win):
        nwin = 1
        frame_len = int(win)
    else:
        win = np.asarray(win)
        nwin = len(win)
        frame_len = nwin

    # Increment/hop
    if inc is None:
        inc = frame_len
    inc = int(inc)

    # Number of frames
    nf = int(np.floor((nx - frame_len + inc) / inc))
    f = np.zeros((nf, frame_len))

    # Frame indices
    for i in range(nf):
        start = i * inc
        f[i, :] = x[start:start + frame_len]

    # Apply window if provided as array
    if nwin > 1:
        w = win.flatten()
        f = f * w[np.newaxis, :]

    return f
```

```python
import numpy as np

def ditherit(x, db=-1, mode='db'):
    """
    Dither a signal by adding very low-level random noise.

    Parameters
    ----------
    x : np.ndarray
        Input signal array.
    db : float, optional
        Dithering amount. For 'db' mode, this is in decibels (default: -96 dB).
        For 'bit' mode, this is treated as the amplitude range (default: 1).
    mode : str, optional
        Type of dithering: 'db' or 'bit' (default: 'db').

    Returns
    -------
    x_dithered : np.ndarray
        Signal with dither noise added.
    """

    x = np.asarray(x, dtype=float)

    mode = mode.lower()
    if mode == 'db':
        # Add Gaussian noise scaled by 10^(db/20)
        noise = (10 ** (db / 20.0)) * np.random.randn(*x.shape)
        x = x + noise

    elif mode == 'bit':
        # Add uniform random integer noise in range [-db, db]
        noise = np.round(2 * db * np.random.rand(*x.shape) - db)
        x = x + noise

    else:
        raise ValueError("Unsupported dithering mode. Use 'db' or 'bit'.")

    return x
```

```python
import numpy as np

def ditherit_2(x, db=None, dtype=None, amp=None, mode=None, **kwargs):
    """
    Dithers a signal by db dB or bit amount.
    Backward-compatible with older code that uses 'amp' or 'mode'.
    """
    # Handle aliases
    if db is None:
        if amp is not None:
            db = amp
        else:
            db = -1
    if dtype is None:
        if mode is not None:
            dtype = mode
        else:
            dtype = 'db'

    x = np.asarray(x, dtype=float)

    if dtype.lower() == 'db':
        noise = (10 ** (db / 20)) * np.random.randn(*x.shape)
        x = x + noise
    elif dtype.lower() == 'bit':
        noise = np.round(2 * db * np.random.rand(*x.shape) - db)
        x = x + noise
    else:
        raise ValueError("Invalid dtype. Use 'db' or 'bit'.")

    return x
```

```python
import numpy as np
from scipy.signal.windows import hamming
#from scipy.signal import lfilter


def fdlp_env_comp_100hz(x, sr=8000, num_ceps=14, flag_delta=0, do_gain_norm=0, nochan=1):
    """
    Python version of fdlp_env_comp_100hz.m
    Computes FDLP envelope-based features at 100Hz frame rate.

    Parameters:
        x: np.ndarray — input signal (1D or 2D, row vector)
        sr: int — sampling rate
        num_ceps: int — number of cepstral coefficients (unused directly here)
        flag_delta: int — include temporal delta features (default=0)
        do_gain_norm: int — gain normalization flag
        nochan: int — number of channels

    Returns:
        feats: np.ndarray — feature matrix [bands x frames]
    """

    if x.ndim > 1 and x.shape[0] > x.shape[1]:
        x = x.T
    x = np.asarray(x).flatten()
    ##bp()

    # ---------------------------------------------------------------------
    # Parameters
    # ---------------------------------------------------------------------
    dB = 48
    flen = int(0.025 * sr)  # 25 ms window
    fhop = int(0.010 * sr)  # 10 ms hop

    padlen = 0
    fdlplen = len(x)

    # Mirror-padding (same as fliplr in MATLAB)
    if padlen > 0:
        x = np.concatenate([
            np.flip(x[:fhop * padlen]),
            x,
            np.flip(x[-fhop * padlen:])
        ])

    fnum = int(np.floor((len(x) - flen) / fhop) + 1)
    send = (fnum - 1) * fhop + flen

    # Downsample factor (FDLP modulation scale)
    factor = 40
    flen = int(np.floor(0.025 * sr / factor))
    fhop = int(np.floor(0.010 * sr / factor))

    trap = 10
    mirr_len = trap * fhop
    fdlpwin = int(np.floor((0.225 * sr) / factor))
    fdlpolap = fdlpwin - fhop
    min_len = 0.11 * sr

    # ---------------------------------------------------------------------
    # FDLP Envelope Computation
    # ---------------------------------------------------------------------
    npts = int(np.floor(fdlplen / factor))

    #bp()

    ENV = fdlpfit_full_sig_vAR(x, sr, dB, do_gain_norm, 'mel', npts, nochan)
    ENV = np.array(ENV)

    nb = ENV.shape[1]  # number of subbands
    start_band = 0

    feats_old = ENV.T
    ENV = ENV.T

    fdlp_spec = ENV[:, :int(np.floor(send / factor))]

    # ---------------------------------------------------------------------
    # Compute short-term energy in each subband
    # ---------------------------------------------------------------------
    #fnum = int(np.floor((len(x) - flen) / fhop) + 1)
    bandenergy = np.zeros((nb, fnum))
    wind = hamming(flen, sym=False)

    for band in range(nb):
        banddata = frame_signal(fdlp_spec[band, :], flen, flen - fhop)
        #bp()
        bandenergy[band, :] = np.dot(banddata, wind)

    ENV = bandenergy ** 0.10  # amplitude compression

    # ---------------------------------------------------------------------
    # Optional frequency delta
    # ---------------------------------------------------------------------
    flag_freq_delta = 0
    if flag_freq_delta == 1:
        del_mat = np.zeros_like(ENV)
        for j in range(nb):
            if j == 0:
                del_mat[j, :] = -ENV[j, :] + ENV[j + 1, :]
            elif j == nb - 1:
                del_mat[j, :] = -ENV[j - 1, :] + ENV[j, :]
            else:
                del_mat[j, :] = -ENV[j - 1, :] + ENV[j + 1, :]
        feats = np.vstack((ENV, del_mat))
    else:
        feats = ENV

    # ---------------------------------------------------------------------
    # Unpadding
    # ---------------------------------------------------------------------
    if padlen > 0:
        feats = feats[:, fhop * padlen: -fhop * padlen]

    return feats


def frame_signal(x, frame_len, overlap):
    """
    Equivalent to MATLAB buffer(x, frame_len, overlap, 'nodelay')
    Creates overlapping frames of a 1D signal.
    """
    step = frame_len - overlap
    num_frames = int(np.floor((len(x) - overlap) / step))
    shape = (num_frames, frame_len)
    strides = (x.strides[0] * step, x.strides[0])
    frames = np.lib.stride_tricks.as_strided(x, shape=shape, strides=strides)
    return frames
```

```python
import numpy as np
from scipy.fftpack import dct

def fdlpfit_full_sig_vAR(x, sr=8000, dB=48, do_gain_norm=False, band_decom='mel', npts=None, nochan=1):
    """
    Fit FDLP model to a signal using sub-band decomposition and variable-order AR.

    Parameters
    ----------
    x : np.ndarray
        Input signal, shape (nochan, samples)
    sr : int
        Sampling rate
    dB : int
        Dynamic range for envelope computation
    do_gain_norm : bool
        Normalize gain if True
    band_decom : str
        Type of sub-band decomposition: 'bark', 'mel', 'oct', 'gammatone', 'uniform'
    npts : int
        Number of points in final envelope
    nochan : int
        Number of channels

    Returns
    -------
    ENVall : np.ndarray
        Envelope features, shape (npts, nb*nochan)
    """

    x = np.atleast_2d(x)
    nchan, flen = x.shape

    ##bp()

    # DCT of the input signal (over samples)
    x_dct = dct(x, axis=1, norm='ortho')

    # -------------------------
    # Sub-band decomposition
    # -------------------------
    if band_decom.lower() == 'mel':
        wts, idx = melweights_mod(flen, sr)
        ##bp()
        factor = 300
        fp = max(round(flen / factor), 1)
        wts, idx = bandlimit_wtsremoval(wts, idx, flen, sr, 0, 8000) # Edited by Anurenjan on 29-03-26 based on the reququirement from Ameen
    elif band_decom.lower() == 'bark':
        wts, idx = barkweights(flen, sr)
        fp = round(flen / 100)
    else:
        # uniform or other
        nbands = min(96, round(flen / 100))
        wts, idx = unif_rect_wind_fixed(nbands, flen)
        fp = max(round((idx[0][1]-idx[0][0])/6), 1)

    nb = len(idx)
    if npts is None:
        npts = flen

    ENVall = np.zeros((nb * nochan, npts))

    # -------------------------
    # Compute envelope per band
    # -------------------------
    K = 1
    for I in range(0, nb, K):
        currBands = list(range(I, min(I + K, nb)))
        arrCurrBands = []

        # Merge sub-bands
        lenCurrBands = min([idx[b][1] - idx[b][0] for b in currBands])

        for b in currBands:
            ##bp()
            temp = x_dct[:, idx[b][0]:idx[b][1]+1] * wts[b][np.newaxis, :]
            temp = temp[:, :lenCurrBands]
            arrCurrBands.append(temp)

        arrCurrBands = np.vstack(arrCurrBands)
        # Envelope extraction using vAR
        #bp()
        tempENV = env_hlpc_vAR(arrCurrBands, fp, npts, do_gain_norm)
        start_idx = (I // K) * K * nochan
        ENVall[start_idx:start_idx + K * nochan, :] = tempENV

    return ENVall.T
```

```python
import numpy as np

def hz2mel(f, htk=False):
    """
    Convert frequencies (in Hz) to mel scale.

    Parameters
    ----------
    f : float or np.ndarray
        Input frequency or array of frequencies (Hz)
    htk : bool, optional
        If True, use the HTK mel scale definition.
        Otherwise, use Slaney's formula (default: False)

    Returns
    -------
    z : float or np.ndarray
        Mel-scaled frequency
    """
    f = np.asanyarray(f)

    if htk:
        # HTK formula
        z = 2595 * np.log10(1 + f / 700.0)
    else:
        # Slaney's Auditory Toolbox definition
        f_0 = 0.0
        f_sp = 200.0 / 3.0  # 66.6667
        brkfrq = 1000.0
        brkpt = (brkfrq - f_0) / f_sp
        logstep = np.exp(np.log(6.4) / 27.0)  # ≈ 1.07117

        # initialize output
        z = np.zeros_like(f)

        # linear region
        linpts = f < brkfrq
        z[linpts] = (f[linpts] - f_0) / f_sp

        # log region
        z[~linpts] = brkpt + np.log(f[~linpts] / brkfrq) / np.log(logstep)

    return z
```

```python
import numpy as np

def melweights_mod(flen, sr, dB=48):
    """
    Generate Gaussian Mel filter weights for FDLP.

    Parameters
    ----------
    flen : int
        Length of the DCT/FFT (number of bins)
    sr : int
        Sampling rate
    dB : float
        Dynamic range threshold (default 48 dB)

    Returns
    -------
    wts : list of np.ndarray
        List of Gaussian weights per Mel band
    idx : np.ndarray
        Start and end indices of bins used for each band (nbands x 2)
    """

    nfreqs = flen

    ##bp()

    # Number of Mel bands
    maxmel = hz2mel(sr / 2)
    #nbands = 42  # hard-coded Commented by Anurenjan on 29-03-26 based on the reququirement from Ameen
    nbands = 128  # hard-coded made 128 by Anurenjan on 29-03-26 based on the reququirement from Ameen

    step = maxmel / (nbands - 1)

    # Mel frequency of every DCT/FFT bin
    freqs_hz = np.linspace(0, sr / 2, nfreqs)
    bin_mel = hz2mel(freqs_hz)

    # Linear scale threshold
    lin = 10 ** (-dB / 20)

    # Allocate
    wts = []
    idx = np.zeros((nbands, 2), dtype=int)

    for i in range(nbands):
        f_mel_mid = i * step
        weight = np.exp(-0.5 * ((bin_mel - f_mel_mid) ** 2))
        # Find indices above threshold
        valid_idx = np.where(weight >= lin)[0]
        if len(valid_idx) == 0:
            valid_idx = np.array([0, 0])
        idx[i, 0] = valid_idx[0]
        idx[i, 1] = valid_idx[-1]
        wts.append(weight[valid_idx])

    return wts, idx
```

```python
import numpy as np

def bandlimit_wtsremoval(wts, idx, flen, sr, minfreq, maxfreq):
    """
    Band-limit the weights by removing sub-bands outside a frequency range.

    Parameters
    ----------
    wts : list of np.ndarray
        List of weight arrays per band
    idx : np.ndarray
        Indices corresponding to each weight array (nbands x 2)
    flen : int
        Length of the frame (DCT or FFT bins)
    sr : int
        Sampling rate
    minfreq : float
        Minimum allowed frequency (Hz)
    maxfreq : float
        Maximum allowed frequency (Hz)

    Returns
    -------
    new_wts : list of np.ndarray
        Filtered list of weight arrays
    new_idx : np.ndarray
        Filtered indices array (nbands_filtered x 2)
    """

    new_wts = []
    new_idx = []

    for i in range(len(wts)):
        tmpx = np.zeros(flen)
        start, end = idx[i]
        tmpx[start:end+1] = wts[i]

        # Find peak
        peak_idx = np.argmax(tmpx)
        peak_freq = peak_idx * (sr / 2) / flen

        # Keep only if peak is inside desired frequency range
        if minfreq <= peak_freq <= maxfreq:
            new_wts.append(wts[i])
            new_idx.append([start, end])

    new_idx = np.array(new_idx, dtype=int)
    return new_wts, new_idx
```

```python
import numpy as np
from statsmodels.tsa.api import VAR

def env_hlpc_vAR(Y, nAR, N, do_gain_norm=True):
    """
    Vector AR modeling to generate sub-band envelopes.

    Parameters
    ----------
    Y : np.ndarray
        Input signal array (channels x time). Typical: sub-band DCT along rows
    nAR : int
        AR order
    N : int
        Number of output samples for envelope
    do_gain_norm : bool
        Whether to apply gain normalization

    Returns
    -------
    ENV : np.ndarray
        Envelope output of shape (channels x N)
    """

    #bp()

    # Ensure time along columns, channels along rows
    Y = np.atleast_2d(Y)
    if Y.shape[1] > Y.shape[0]:
        Y = Y.T

    n_channels, n_samples = Y.shape

    # Energy of each channel
    #bp()
    E = np.sum(Y**2, axis=0)

    #E = np.sum(Y**2, axis=1)

    # Fit vector AR model (multichannel)
    # statsmodels VAR expects shape (time x channels)
    ##bp()

    w, A, C, Cnorm = arfit_2(Y, nAR, nAR, 'zero')

    #model = VAR(Y.T)
    #results = model.fit(nAR)

    # Coefficients: results.coefs.shape = (nAR, n_channels, n_channels)
    # Noise covariance: results.sigma_u
    #A = results.coefs  # AR matrices
    #C = results.sigma_u  # Noise covariance

    # Generate envelope using AR coefficients
    ##bp()
    ENV = estimate_mvar_ARFIT_fast_2(A, C, N, E, do_gain_norm)

    return ENV
```

```python
import numpy as np

def arfit_2(v, pmin, pmax=None, selector=None):
    """
    MATLAB-compatible ARFIT for the fixed-order case (pmin == pmax).
    Usage: [w, A, C, Cnorm] = arfit_2(Y, nAR, nAR, 'zero')
    """

    # Handle MATLAB-style args
    if pmax is None:
        p = int(pmin)
    else:
        if pmin != pmax:
            raise ValueError("This version supports only pmin == pmax case.")
        p = int(pmin)

    no_const = False
    if isinstance(selector, str) and selector.lower() == 'zero':
        no_const = True

    v = np.asarray(v)
    if v.ndim == 1:
        v = v.reshape(-1, 1)

    n, m = v.shape

    # Create lag matrix
    Z = np.zeros((n - p, m * p))
    for j in range(p):
        Z[:, j*m:(j+1)*m] = v[p - j - 1:n - j - 1, :]

    if not no_const:
        Z = np.hstack((Z, np.ones((n - p, 1))))

    Y = v[p:, :]

    # Solve least squares
    coef, _, _, _ = np.linalg.lstsq(Z, Y, rcond=None)
    E = Y - Z @ coef  # residuals

    if not no_const:
        w = coef[-1, :].T
        A = coef[:-1, :].T
        mcor = 1
    else:
        w = np.zeros((1, m))
        A = coef.T
        mcor = 0

    ne = n - p
    np_param = m * p + mcor
    dof = ne - np_param
    if dof <= 0:
        raise ValueError(f"Degrees of freedom <= 0 (dof={dof}). Time series too short for p={p}.")

    # Covariance
    C = (E.T @ E) / dof

    detC = np.linalg.det(C)
    if detC <= 0:
        Cnorm = C
    else:
        Cnorm = C / (detC ** (1.0 / m))

    return w, A, C, Cnorm
```

```python
import numpy as np

def estimate_mvar_ARFIT_fast_2(AModel, Q, N, E=None, do_gain_norm=True):
    """
    Python equivalent of MATLAB estimate_mvar_ARFIT_fast.
    """
    d = Q.shape[0]  # number of channels
    p = AModel.shape[1] // d  # AR order

    I = np.eye(d)
    AR = np.hstack([I, -AModel])  # [I, -A1, -A2, ... -Ap]

    # --- MATLAB reshape equivalence (column-major) ---
    AR_mod = np.reshape(AR, (d * d, p + 1), order='F')

    # --- MATLAB fft along 2nd dimension (dim=2) ---
    A_fft = np.fft.fft(AR_mod, n=2 * N - 1, axis=1)
    A_fft = A_fft[:, :N]

    ##bp()

    x = np.zeros((d, N))

    for n in range(N):
        # MATLAB uses H = inv(reshape(A(:,n), d, d))
        H = np.linalg.inv(A_fft[:, n].reshape((d, d), order='F'))
        x[:, n] = 2 * np.real(np.diag(H @ Q @ H.conj().T))

    ##bp()

    if do_gain_norm:
        x = x / np.sum(x, axis=1, keepdims=True)

    return x
```

```python
import numpy as np
import librosa

def melfcc_python(samples, sr=16000, wintime=0.025, hoptime=0.010,
                  numcep=13, preemph=0.97, minfreq=200, maxfreq=6500,
                  nbands=36, lifterexp=0.6, sumpower=True):
    """
    Approximate implementation of MATLAB's melfcc() using librosa.
    """
    # 1. Pre-emphasis
    if preemph != 0:
        samples = np.append(samples[0], samples[1:] - preemph * samples[:-1])

    # 2. STFT -> Power spectrum
    nfft = int(sr * wintime)
    hop_length = int(sr * hoptime)
    S = np.abs(librosa.stft(samples, n_fft=nfft, hop_length=hop_length))**2

    # 3. Mel filterbank
    mel_basis = librosa.filters.mel(sr=sr, n_fft=nfft, n_mels=nbands,
                                    fmin=minfreq, fmax=maxfreq)
    mel_spec = np.dot(mel_basis, S)

    # 4. Log compression
    log_mel_spec = np.log(np.maximum(mel_spec, 1e-10))

    # 5. DCT -> cepstra
    cepstra = librosa.feature.mfcc(S=log_mel_spec, n_mfcc=numcep, dct_type=2)

    # 6. Liftering (fix broadcasting issue)
    if lifterexp != 0:
        n = np.arange(cepstra.shape[0])
        lifter = 1 + (cepstra.shape[0] / 2) * (np.sin(np.pi * n / cepstra.shape[0])) ** lifterexp
        cepstra *= lifter[:, np.newaxis]  # reshape to (numcep, 1)

    # 7. Power scaling
    aspectrum = mel_spec if sumpower else np.sqrt(mel_spec)

    return cepstra, aspectrum, S
```

```python
import matplotlib.pyplot as plt
import librosa.display
import numpy as np

# Compute cepstra and mel spectrum
y, sr = librosa.load("/content/drive/MyDrive/FDLP/c3rc020n.wav", sr=16000)
cepstra, aspectrum, pspectrum = melfcc_python(y, sr)

# Frame indices
frames = np.arange(cepstra.shape[1])

# Mel frequencies corresponding to mel bands
mel_freqs = librosa.mel_frequencies(
    n_mels=aspectrum.shape[0], fmin=200, fmax=6500
)

plt.figure(figsize=(14, 6))

# --- Plot 1: Cepstra ---
plt.subplot(1, 2, 1)
plt.imshow(cepstra, aspect='auto', origin='lower', cmap='jet',
           extent=[0, frames[-1], mel_freqs[0], mel_freqs[-1]])
plt.title('Cepstra (MFCCs)')
plt.xlabel('Frame Index')
plt.ylabel('Frequency (Hz)')
plt.colorbar(label='Amplitude')

# --- Plot 2: Mel Spectrum ---
plt.subplot(1, 2, 2)
plt.imshow(np.log(aspectrum + 1e-10), aspect='auto', origin='lower', cmap='jet',
           extent=[0, frames[-1], mel_freqs[0], mel_freqs[-1]])
plt.title('Mel Filterbank Spectrum (Log Scale)')
plt.xlabel('Frame Index')
plt.ylabel('Frequency (Hz)')
plt.colorbar(label='Log Energy')

plt.tight_layout()
plt.show()
```
